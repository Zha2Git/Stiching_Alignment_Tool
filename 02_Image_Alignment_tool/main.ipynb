## Notebook to run the core function in Alignment tool

```python

# Run main.py core functionality directly in the Notebook
No command line needed; set parameters and call the main processing function.

Parameters:
- `source`: Input folder path containing raw images (e.g., TIF files).
- `destination`: Output folder path where processed images are saved.
- `yes`: Execute without interaction (True for automatic run).
- `reference`: Reference channel name used for registration, typically "DAPI".
- `resolution`: Resolution of the raw images (microns per pixel).
- `disable_reassemble`: Skip reassembling CZI files (True to skip).
- `disable_registration`: Skip image registration (True to skip).
- `downscale`: Downsample large images (True to downsample).
- `factor`: Downsampling ratio, only effective when `downscale=True`.
- `nofinalimage`: Do not generate the final image combining all channels (default False, generate final image).
- `background`: Background subtraction parameter, usually 'False' or a specified background filename.
- `backgroundMult`: Background intensity multiplier.
- `fullname`: Whether output image metadata contains full channel names (True to include).
- `pyramidal`: Additionally save pyramid multi-resolution OME-TIFF (True to save).
- `output`: Output file type ('tif').

```
